# 🗺️ Módulo 1: Sistemas de Referencia y Coordenadas en el Plano Cartesiano
**Evidencia:** *2.2 Utilizo sistemas de referencia para representar la ubicación de objetos geométricos.*

### Conceptos:
- **Eje X (Abscisas):** Movimiento horizontal (positivo a la derecha, negativo a la izquierda).
- **Eje Y (Ordenadas):** Movimiento vertical (positivo hacia arriba, negativo hacia abajo).
- **Origen $(0,0)$:** Punto central donde se cruzan ambos ejes.
- **Par ordenado $(x, y)$:** Dirección exacta de un punto en el plano.
- **Polígono 2D:** Figura cerrada delimitada por vértices coordenados.

```python
%%manim -qm -v WARNING PlanoCartesianoYPosicion

class PlanoCartesianoYPosicion(Scene):
    def construct(self):
        titulo = Text("1. Sistema de Referencia: El Plano Cartesiano", font_size=30, color="#00D2FF")
        subtitulo = Text("Ubicación de puntos y figuras 2D mediante coordenadas (x, y)", font_size=18, color="#FFD166")
        cabecera = VGroup(titulo, subtitulo).arrange(DOWN, aligned_edge=LEFT, buff=0.12).to_corner(UL, buff=0.4)
        self.play(Write(titulo), FadeIn(subtitulo), run_time=1.0)

        plano = NumberPlane(
            x_range=[-5, 5, 1], y_range=[-3, 3, 1],
            x_length=8.5, y_length=5.1,
            background_line_style={"stroke_color": "#2A364F", "stroke_width": 1.5},
            axis_config={"stroke_color": "#8D99AE", "stroke_width": 2.5, "include_numbers": True, "font_size": 18}
        ).shift(DOWN * 0.4 + RIGHT * 0.5)

        etiqueta_x = MathTex("X", color="#00D2FF", font_size=24).next_to(plano.x_axis.get_end(), RIGHT, buff=0.15)
        etiqueta_y = MathTex("Y", color="#00D2FF", font_size=24).next_to(plano.y_axis.get_end(), UP, buff=0.15)
        self.play(Create(plano), Write(etiqueta_x), Write(etiqueta_y), run_time=1.2)

        origen_punto = Dot(plano.c2p(0, 0), color="#FFD166", radius=0.08)
        origen_texto = Text("Origen (0, 0)", font_size=16, color="#FFD166").next_to(origen_punto, DL, buff=0.12)
        self.play(FadeIn(origen_punto), Write(origen_texto), run_time=0.6)

        coords = [("A", 2, 2, "#FF6B6B"), ("B", -3, 1, "#06D6A0"), ("C", -2, -2, "#FFD166"), ("D", 3, -1, "#00D2FF")]
        puntos_dots, puntos_labels, lineas_guias = [], [], []
        for nom, x, y, col in coords:
            pos = plano.c2p(x, y)
            dot = Dot(pos, color=col, radius=0.09)
            lbl = MathTex(f"{nom}({x}, {y})", font_size=20, color=col).next_to(dot, UR if x>=0 else UL, buff=0.12)
            guia = VGroup(
                DashedLine(plano.c2p(x, 0), pos, color=col, stroke_width=1.5, stroke_opacity=0.7),
                DashedLine(plano.c2p(0, y), pos, color=col, stroke_width=1.5, stroke_opacity=0.7)
            )
            puntos_dots.append(dot)
            puntos_labels.append(lbl)
            lineas_guias.append(guia)

        for dot, lbl, guia in zip(puntos_dots, puntos_labels, lineas_guias):
            self.play(Create(guia), FadeIn(dot, scale=1.3), Write(lbl), run_time=0.6)

        vertices = [plano.c2p(x, y) for _, x, y, _ in coords]
        poligono = Polygon(*vertices, color="#FFD166", stroke_width=3.5, fill_color="#FFD166", fill_opacity=0.25)
        lbl_fig = Text("Polígono 2D (4 vértices conectados)", font_size=18, color="#FFD166").to_corner(DL, buff=0.4)
        self.play(Create(poligono), Write(lbl_fig), run_time=1.2)
        self.wait(1.0)
```

# 🔄 Módulo 2: Movimientos Rígidos en el Plano (Isometrías)
**Competencia:** *1. Reconoce las características medibles y de posición de objetos bidimensionales y de movimientos simples de estos: rotación, traslación y reflexión.*

### Tres tipos de movimientos rígidos:
1. **Traslación:** Desplazamiento por vector $\vec{v} = (a, b)$.
2. **Rotación:** Giro respecto a un punto central $O$ con ángulo $\theta$.
3. **Reflexión:** Espejo a través de un eje de simetría (invierte la orientación).

```python
%%manim -qm -v WARNING TransformacionesIsometricas

class TransformacionesIsometricas(Scene):
    def construct(self):
        titulo = Text("2. Movimientos Rígidos en el Plano", font_size=30, color="#00D2FF")
        subtitulo = Text("Conservan las medidas, forma y área exacta de la figura", font_size=18, color="#A0AABF")
        cabecera = VGroup(titulo, subtitulo).arrange(DOWN, aligned_edge=LEFT, buff=0.1).to_corner(UL, buff=0.35)
        self.play(Write(titulo), FadeIn(subtitulo), run_time=1.0)

        plano = NumberPlane(
            x_range=[-6, 6, 1], y_range=[-3.5, 3.5, 1],
            x_length=11.5, y_length=5.5,
            background_line_style={"stroke_color": "#1F293D", "stroke_width": 1.2}
        ).shift(DOWN * 0.4)
        self.play(Create(plano), run_time=0.8)

        p1, p2, p3 = plano.c2p(-4, -1), plano.c2p(-2, -1), plano.c2p(-2.5, 1)
        triangulo_orig = Polygon(p1, p2, p3, color="#00D2FF", fill_color="#00D2FF", fill_opacity=0.4, stroke_width=3)
        lbl_orig = Text("Original", font_size=16, color="#00D2FF").next_to(triangulo_orig, DOWN, buff=0.15)
        self.play(Create(triangulo_orig), Write(lbl_orig), run_time=0.8)

        # Traslación
        banner_trans = Text("A) TRASLACIÓN: Vector (+3, +1)", font_size=20, color="#06D6A0").to_corner(UR, buff=0.4)
        flecha_v = Arrow(plano.c2p(-4, -1), plano.c2p(-1, 0), color="#06D6A0", buff=0, stroke_width=3)
        triangulo_tras = triangulo_orig.copy().set_color("#06D6A0").set_fill("#06D6A0", opacity=0.4)
        self.play(Write(banner_trans), Create(flecha_v))
        self.play(triangulo_tras.animate.shift(plano.c2p(3, 1) - plano.c2p(0, 0)), run_time=1.2)
        self.wait(0.5)
        self.play(FadeOut(flecha_v), FadeOut(triangulo_tras), FadeOut(banner_trans), run_time=0.5)

        # Rotación
        banner_rot = Text("B) ROTACIÓN: 90° alrededor del Origen", font_size=20, color="#FFD166").to_corner(UR, buff=0.4)
        centro = Dot(plano.c2p(0, 0), color="#FFD166", radius=0.08)
        triangulo_rot = triangulo_orig.copy().set_color("#FFD166").set_fill("#FFD166", opacity=0.4)
        self.play(Write(banner_rot), FadeIn(centro))
        self.play(Rotate(triangulo_rot, angle=PI/2, about_point=plano.c2p(0, 0)), run_time=1.5)
        self.wait(0.5)
        self.play(FadeOut(centro), FadeOut(triangulo_rot), FadeOut(banner_rot), run_time=0.5)

        # Reflexión
        banner_ref = Text("C) REFLEXIÓN: Simetría axial (Eje Y)", font_size=20, color="#FF6B6B").to_corner(UR, buff=0.4)
        eje_espejo = Line(plano.c2p(0, -3), plano.c2p(0, 3), color="#FF6B6B", stroke_width=3.5)
        p1_r, p2_r, p3_r = plano.c2p(4, -1), plano.c2p(2, -1), plano.c2p(2.5, 1)
        triangulo_ref = Polygon(p1_r, p2_r, p3_r, color="#FF6B6B", fill_color="#FF6B6B", fill_opacity=0.4, stroke_width=3)
        self.play(Write(banner_ref), Create(eje_espejo))
        self.play(TransformFromCopy(triangulo_orig, triangulo_ref), run_time=1.5)
        self.wait(1.0)
```

# 📏 Módulo 3: Atributos Medibles: Longitud (1D) vs Área (2D)
**Evidencias:**
- *1.1 Señala los atributos medibles de una figura junto con sus posibles unidades y magnitudes.*
- *2.3 Reconozco el conjunto de unidades usadas para cada magnitud: longitud y área.*

| Característica | Perímetro (Longitud - 1D) | Área (Superficie - 2D) |
|---|---|---|
| **Definición** | Medida del contorno exterior | Medida del recubrimiento interior |
| **Dimensión** | 1 Dimensión (Línea continua) | 2 Dimensiones (Cuadrículas de recubrimiento) |
| **Unidades SI** | $\text{mm, cm, m, km}$ | $\text{mm}^2, \text{cm}^2, \text{m}^2, \text{km}^2$ |

```python
%%manim -qm -v WARNING LongitudVsArea

class LongitudVsArea(Scene):
    def construct(self):
        titulo = Text("3. Atributos Medibles: Longitud (1D) vs Área (2D)", font_size=30, color="#00D2FF")
        subtitulo = Text("Perímetro = Contorno desenrollado | Área = Cuadrículas interiores", font_size=18, color="#FFD166")
        cabecera = VGroup(titulo, subtitulo).arrange(DOWN, aligned_edge=LEFT, buff=0.12).to_corner(UL, buff=0.35)
        self.play(Write(titulo), FadeIn(subtitulo), run_time=1.0)

        w, h = 4.0 * 0.75, 3.0 * 0.75
        rect = Rectangle(width=w, height=h, color="#00D2FF", stroke_width=4).shift(UP * 0.2 + LEFT * 3.0)
        
        lbl_top = MathTex("4\\text{ m}", font_size=20, color="#00D2FF").next_to(rect, UP, buff=0.12)
        lbl_bot = MathTex("4\\text{ m}", font_size=20, color="#00D2FF").next_to(rect, DOWN, buff=0.12)
        lbl_izq = MathTex("3\\text{ m}", font_size=20, color="#00D2FF").next_to(rect, LEFT, buff=0.12)
        lbl_der = MathTex("3\\text{ m}", font_size=20, color="#00D2FF").next_to(rect, RIGHT, buff=0.12)
        self.play(Create(rect), Write(VGroup(lbl_top, lbl_bot, lbl_izq, lbl_der)), run_time=1.0)

        tit_perim = Text("A) PERÍMETRO (Longitud - 1D)", font_size=20, color="#06D6A0").to_corner(UR, buff=0.4)
        form_perim = MathTex(r"P = 4 + 3 + 4 + 3 = 14\text{ m}", font_size=22, color="#06D6A0").next_to(tit_perim, DOWN, aligned_edge=LEFT, buff=0.2)
        self.play(Write(tit_perim), Write(form_perim), run_time=1.0)

        linea_recta = Line(LEFT * 5.0, RIGHT * 2.0, color="#06D6A0", stroke_width=5).to_edge(DOWN, buff=0.8)
        lbl_lin = MathTex(r"\text{Línea recta continua: } 14\text{ metros}", font_size=20, color="#06D6A0").next_to(linea_recta, UP, buff=0.1)
        self.play(Create(linea_recta), Write(lbl_lin), run_time=1.2)

        tit_area = Text("B) ÁREA (Superficie - 2D)", font_size=20, color="#FFD166").next_to(form_perim, DOWN, aligned_edge=LEFT, buff=0.4)
        form_area = MathTex(r"A = 4\text{ m} \times 3\text{ m} = 12\text{ m}^2", font_size=22, color="#FFD166").next_to(tit_area, DOWN, aligned_edge=LEFT, buff=0.2)
        self.play(Write(tit_area), Write(form_area), run_time=1.0)

        dx, dy = w / 4.0, h / 3.0
        tl = rect.get_corner(UL)
        bl = rect.get_corner(DL)
        cuadritos = VGroup()
        for i in range(4):
            for j in range(3):
                c = Square(side_length=min(dx, dy)*0.95, fill_color="#FFD166", fill_opacity=0.5, stroke_color="#FFD166", stroke_width=1.5)
                c.move_to([tl[0] + (i + 0.5)*dx, bl[1] + (j + 0.5)*dy, 0])
                cuadritos.add(c)

        self.play(LaggedStart(*[FadeIn(c, scale=0.3) for c in cuadritos], lag_ratio=0.06), run_time=1.5)
        self.wait(1.0)
```

# 📦 Módulo 4: Relación entre Figuras 2D y Sólidos 3D
**Evidencia:** *2.1 Identifico relaciones entre figuras bidimensionales y tridimensionales.*

Un sólido tridimensional ($3\text{D}$) se construye a partir de una plantilla o red plana bidimensional ($2\text{D}$):
- **Cubo:** 6 caras cuadradas congruentes, 12 aristas y 8 vértices.
- **Relación de Euler:** $\text{Caras} + \text{Vértices} - \text{Aristas} = 2$ ($6 + 8 - 12 = 2$).

```python
%%manim -qm -v WARNING De2Da3DDesdoblamiento

class De2Da3DDesdoblamiento(ThreeDScene):
    def construct(self):
        titulo = Text("4. Relación 2D y 3D: Desarrollo Plano a Sólido", font_size=30, color="#00D2FF")
        subtitulo = Text("Una red 2D de 6 cuadrados se pliega para formar un Cubo (3D)", font_size=18, color="#FFD166")
        cabecera = VGroup(titulo, subtitulo).arrange(DOWN, aligned_edge=LEFT, buff=0.1).to_corner(UL, buff=0.35)
        self.add_fixed_in_frame_mobjects(cabecera)
        self.play(Write(titulo), FadeIn(subtitulo), run_time=1.0)

        L = 1.5
        colores = ["#00D2FF", "#FF6B6B", "#06D6A0", "#FFD166", "#9B5DE5", "#F15BB5"]
        nombres = ["Base", "Frente", "Atrás", "Izq", "Der", "Tapa"]
        posiciones_2d = [[0, 0, 0], [0, -L, 0], [0, L, 0], [-L, 0, 0], [L, 0, 0], [0, 2*L, 0]]

        caras_2d = VGroup()
        for pos, col, nom in zip(posiciones_2d, colores, nombres):
            sq = Square(side_length=L, fill_color=col, fill_opacity=0.6, stroke_color=WHITE, stroke_width=2.5).move_to(pos)
            lbl = Text(nom, font_size=14, color=WHITE).move_to(sq.get_center())
            caras_2d.add(VGroup(sq, lbl))

        caras_2d.shift(DOWN * 0.5 + LEFT * 1.5)
        banner_2d = Text("Red 2D (Plantilla plana de 6 caras)", font_size=20, color=WHITE).to_corner(UR, buff=0.5)
        self.add_fixed_in_frame_mobjects(banner_2d)

        self.play(LaggedStart(*[FadeIn(c, scale=0.8) for c in caras_2d], lag_ratio=0.12), Write(banner_2d), run_time=1.5)
        self.wait(0.5)
        self.play(FadeOut(banner_2d), run_time=0.4)

        self.move_camera(phi=65 * DEGREES, theta=-45 * DEGREES, zoom=0.85, run_time=1.5)

        cubo_3d = VGroup(
            Square(side_length=L, fill_color=colores[0], fill_opacity=0.7, stroke_color=WHITE).move_to([0, 0, -L/2]),
            Square(side_length=L, fill_color=colores[1], fill_opacity=0.7, stroke_color=WHITE).rotate(PI/2, axis=RIGHT).move_to([0, -L/2, 0]),
            Square(side_length=L, fill_color=colores[2], fill_opacity=0.7, stroke_color=WHITE).rotate(PI/2, axis=RIGHT).move_to([0, L/2, 0]),
            Square(side_length=L, fill_color=colores[3], fill_opacity=0.7, stroke_color=WHITE).rotate(PI/2, axis=UP).move_to([-L/2, 0, 0]),
            Square(side_length=L, fill_color=colores[4], fill_opacity=0.7, stroke_color=WHITE).rotate(PI/2, axis=UP).move_to([L/2, 0, 0]),
            Square(side_length=L, fill_color=colores[5], fill_opacity=0.7, stroke_color=WHITE).move_to([0, 0, L/2])
        ).shift(RIGHT * 1.0)

        banner_3d = Text("Sólido 3D Plegado (Cubo)", font_size=20, color="#06D6A0").to_corner(UR, buff=0.5)
        self.add_fixed_in_frame_mobjects(banner_3d)

        self.play(ReplacementTransform(caras_2d, cubo_3d), Write(banner_3d), run_time=2.0)
        self.begin_ambient_camera_rotation(rate=0.4)
        self.wait(2.5)
        self.stop_ambient_camera_rotation()
```
